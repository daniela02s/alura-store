```python
import sys
import os
sys.path.append(os.path.abspath(os.path.join('..', 'scripts')))
```

```python
import carregar_dados
```

```python
dados_consolidados = carregar_dados.carregar_e_consolidar_dados()
```

### ***Métricas de Desempenho***

#### 1. Faturamento Total por Loja

```python
# Agrupa o DF pela coluna 'Loja'
# Seleciona a coluna 'Preço' para cada um desses grupos
# Soma os valores de 'Preço' para cada loja

faturamento_total_por_loja = dados_consolidados.groupby('Loja')['Preço'].sum()
```

```python
# Converte a série de faturamento em um DF e aplica a formatação para moeda brasileira (R$)

df_faturamento = faturamento_total_por_loja.to_frame(name='Faturamento')
df_faturamento['Faturamento'] = df_faturamento['Faturamento'].map('R$ {:,.2f}'.format)
```

```python
print("\nFaturamento Total por Loja (R$):\n")
print(df_faturamento)
```

#### 2. Vendas por Categoria

```python
# Lista de lojas únicas para iterar sobre elas
lojas = sorted(dados_consolidados['Loja'].unique())
```

```python
for loja in lojas:
    print(f"\nTOP 5 CATEGORIAS MAIS VENDIDAS - {loja.upper()}\n")
    
    # Filtra o DF para a loja atual
    df_loja = dados_consolidados[dados_consolidados['Loja'] == loja]
    
    # Conta a ocorrência de cada categoria de produto e pega as 5 mais populares
    top_5_categorias = df_loja['Categoria do Produto'].value_counts().head(5)
    
    print(top_5_categorias)
    print("\n")
```

#### 3. Média de Avaliação por Loja

```python
avaliacao_media_por_loja = dados_consolidados.groupby('Loja')['Avaliação da compra'].mean()
```

```python
df_avaliacoes = avaliacao_media_por_loja.to_frame(name='Avaliação Média')

# Formata os valores para exibir com duas casas decimais
df_avaliacoes['Avaliação Média'] = df_avaliacoes['Avaliação Média'].map('{:.2f}'.format)
```

```python
print("\nAvaliação Média dos Clientes por Loja:\n")
print(df_avaliacoes)
```

#### 3. Produtos Mais e Menos Vendidos

```python
lojas = sorted(dados_consolidados['Loja'].unique())
```

```python
for loja in lojas:
    print(f"\nANÁLISE DE PRODUTOS - {loja.upper()}\n")

    df_loja = dados_consolidados[dados_consolidados['Loja'] == loja]

    # Conta a frequência de cada produto
    contagem_produtos = df_loja['Produto'].value_counts()
    
    # Produtos Mais Vendidos
    print(f"\n--- TOP 5 PRODUTOS MAIS VENDIDOS ---\n")
    print(contagem_produtos.head(5))
    
    # Produtos Menos Vendidos
    print(f"\n--- TOP 5 PRODUTOS MENOS VENDIDOS ---\n")
    print(contagem_produtos.tail(5))
    print("\n")
```

#### 5. Frete Médio por Loja

```python
frete_medio_por_loja = dados_consolidados.groupby('Loja')['Frete'].mean()

df_frete = frete_medio_por_loja.to_frame(name='Frete Médio (R$)')

# Aplica a formatação para moeda brasileira (R$)
df_frete['Frete Médio (R$)'] = df_frete['Frete Médio (R$)'].map('R$ {:,.2f}'.format)
```

```python
# Exibe a tabela formatada
print("\nCusto Médio de Frete por Loja:\n")
print(df_frete)
```
