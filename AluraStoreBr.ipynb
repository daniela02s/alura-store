#### 2.1 Faturamento

O faturamento é o principal indicador da saúde financeira de uma loja. A Loja 1 é a líder de receita, enquanto a Loja 4 apresenta o menor faturamento da rede. A diferença de quase R$ 150.000,00 para a primeira colocada evidencia uma performance financeira significativamente inferior, tornando este o ponto mais crítico da sua análise.

#### 2.2 Avaliação dos Clientes

A satisfação do cliente é um indicador vital para a sustentabilidade do negócio. Neste quesito, a Loja 3 (4.05) e a Loja 2 (4.04) são as mais bem avaliadas. Curiosamente, a Loja 1, líder de faturamento, possui a pior avaliação (3.98). A Loja 4, por sua vez, mantém uma avaliação positiva (4.00), indicando que seu baixo desempenho de vendas não está diretamente ligado à insatisfação dos clientes.

#### 2.3 Custo do Frete

O custo do frete pode influenciar a decisão decompra do cliente, a Loja 4 se destaca positivamente por ter o frete médio mais barato para o cliente (R$ 31,28). Isso representa uma vantagem competitiva que, no entanto, não foi suficiente para impulsionar seu faturamento geral. Em contrapartida, a Loja 1, líder em faturamento, possui o frete mais caro (R$ 34,69). Apesar das diferenças, os valores são relativamente próximos, indicando uma política de frete consistente em toda a rede.

O gráfico de dispersão confirma uma forte correlação positiva: produtos mais caros possuem fretes mais altos. A distribuição de pontos coloridos mostra que todas as lojas seguem a mesma política de frete, sem discrepâncias notáveis entre elas. A análise também reforça que a grande maioria das vendas em todas as lojas se concentra em produtos de menor valor (abaixo de R$ 1.000).

#### 2.4 Análise de Produtos e Categorias

A análise detalhada do mix de produtos revela perfis de venda distintos para cada loja:

* **Loja 1:** Focada em produtos de alto valor. Seus itens mais vendidos são eletrodomésticos e móveis de alto custo, como "Micro-ondas", "TV Led UHD 4K" e "Guarda roupas". Curiosamente, seus produtos menos vendidos são itens de baixo valor, como livros e utensílios.

* **Loja 2 e Loja 4:** Apresentam um perfil de vendas mais diversificado. Ambas têm livros entre os itens mais vendidos ("Iniciando em programação" na Loja 2 e "Dashboards com Power BI" na Loja 4), indicando um público interessado em material de estudo.

* **Loja 3:** Possui um foco claro em móveis e utilidades domésticas, com produtos como "Kit banquetas", "Mesa de jantar" e "Jogo de panelas" liderando as vendas.

* **Produtos Menos Vendidos:** A lista de produtos menos vendidos varia muito entre as lojas, mas na Loja 4, itens de alto custo como "Geladeira" e "Guarda roupas" estão entre os de menor saída, o que pode contribuir para seu faturamento geral mais baixo. A loja pode estar falhando em atrair clientes para essas categorias de produtos de margem potencialmente maior, como "móveis".

Essa análise aprofundada mostra que não há uma estratégia única de produtos que funcione para todas as lojas, cada uma atendendo a um perfil de cliente diferente.

A decisão de vender uma unidade deve ser primariamente guiada pelo seu desempenho financeiro. Conforme destacado pelo gráfico de faturamento, a Loja 4 é a que gera a menor receita para a rede, sendo, portanto, a mais ineficiente do ponto de vista de negócios.

Embora a Loja 4 apresente pontos positivos, como o frete competitivo e uma boa avaliação dos clientes, esses fatores não se traduzem em um volume de vendas robusto o suficiente para competir com as outras unidades.
